**This notebook demonstrates building and training a Convolutional Neural Network (CNN) for image classification on the CIFAR-10 dataset. A key aspect of this analysis was the implementation of image enhancement techniques to potentially improve model performance.**

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report
import cv2
from collections import Counter

# Load and prepare the CIFAR-10 dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# CIFAR-10 class names
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

# Visualize a sample of the original dataset
def visualize_sample_images(images, labels, class_names, num_images=10):

    plt.figure(figsize=(10, 10))
    for i in range(num_images):
        plt.subplot(5, 2, i + 1)
        plt.imshow(images[i])
        plt.title(class_names[labels[i][0]])
        plt.axis('off')
    plt.suptitle("Sample Original CIFAR-10 Images", y=1.02)
    plt.tight_layout()
    plt.show()


visualize_sample_images(x_train, y_train, class_names)

# Visualize the composition of the classes
def visualize_class_composition(labels, class_names):

    class_counts = Counter(labels.flatten())
    class_names_sorted = [class_names[i] for i in sorted(class_counts.keys())]
    counts_sorted = [class_counts[i] for i in sorted(class_counts.keys())]

    plt.figure(figsize=(10, 5))
    plt.bar(class_names_sorted, counts_sorted)
    plt.title("CIFAR-10 Class Distribution")
    plt.xlabel("Class")
    plt.ylabel("Number of Images")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()


visualize_class_composition(y_train, class_names)


# Normalize pixel values to [0, 1]
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# Convert labels to categorical
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)



def enhance_image_clarity(images):

    enhanced_images = []
    for img in images:
        # Convert to appropriate format for OpenCV
        img_uint8 = (img * 255).astype(np.uint8)

        # 1. Sharpening filter using kernel
        sharpening_kernel = np.array([[-1,-1,-1],
                                     [-1, 9,-1],
                                     [-1,-1,-1]])
        sharpened = cv2.filter2D(img_uint8, -1, sharpening_kernel)

        # 2. Contrast Limited Adaptive Histogram Equalization (CLAHE)
        # Convert to LAB color space(Luminance separated from A and B)
        lab = cv2.cvtColor(sharpened, cv2.COLOR_RGB2LAB)
        l, a, b = cv2.split(lab)

        # Apply CLAHE to L channel ( To adjust contrast without affecting the color)
        clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
        l_clahe = clahe.apply(l) # Cliplimit helps to prevent the noise from being amplified and title gridsize divides the into smaller

        # Merge back and convert to RGB
        lab_clahe = cv2.merge([l_clahe, a, b])
        enhanced = cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2RGB)

        # 3. Mild Gaussian blur to reduce noise while preserving edges
        enhanced = cv2.GaussianBlur(enhanced, (3,3), 0.5)

        enhanced_images.append(enhanced.astype('float32') / 255.0)

    return np.array(enhanced_images)

# Apply enhancement to training and test data
print("Enhancing training images...")
x_train_enhanced = enhance_image_clarity(x_train)

print("Enhancing test images...")
x_test_enhanced = enhance_image_clarity(x_test)

def create_enhanced_cnn_model():

    model = models.Sequential([
        # First convolutional block with enhanced filters
        layers.Conv2D(32, (3, 3), activation='relu', padding='same',
                     input_shape=(32, 32, 3)),
        layers.BatchNormalization(),
        layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),

        # Second convolutional block
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),

        # Third convolutional block
        layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),

        # Fourth convolutional block for finer details
        layers.Conv2D(256, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.Dropout(0.25),

        # Global Average Pooling instead of Flatten to reduce parameters
        layers.GlobalAveragePooling2D(),

        # Dense layers
        layers.Dense(512, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(10, activation='softmax')
    ])

    return model

# Create and compile the model
model = create_enhanced_cnn_model()

# Custom optimizer with learning rate scheduling
initial_learning_rate = 0.001
lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate,
    decay_steps=1000,
    decay_rate=0.9,
    staircase=True)

optimizer = tf.keras.optimizers.Adam(learning_rate=lr_schedule)

model.compile(optimizer=optimizer,
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Display model architecture
model.summary()

# Callbacks for better training
callbacks = [
    tf.keras.callbacks.EarlyStopping(patience=10, restore_best_weights=True),
]

# Train the model

history = model.fit(
    x_train_enhanced, y_train,
    batch_size=128,
    epochs=100,
    validation_data=(x_test_enhanced, y_test),
    callbacks=callbacks,
    verbose=1
)

# Evaluate the model
test_loss, test_accuracy = model.evaluate(x_test_enhanced, y_test, verbose=0)
print(f"\nTest Accuracy: {test_accuracy:.4f}")
print(f"Test Loss: {test_loss:.4f}")

# Make predictions
y_pred = model.predict(x_test_enhanced)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_test, axis=1)

# Display classification report
print("\nClassification Report:")
print(classification_report(y_true_classes, y_pred_classes,
                          target_names=class_names))

# Visualize original vs enhanced images
def compare_images(original, enhanced, num_images=5):

    fig, axes = plt.subplots(2, num_images, figsize=(15, 6))

    for i in range(num_images):
        # Original image
        axes[0, i].imshow(original[i])
        axes[0, i].set_title(f'Original - {class_names[np.argmax(y_train[i])]}')
        axes[0, i].axis('off')

        # Enhanced image
        axes[1, i].imshow(enhanced[i])
        axes[1, i].set_title('Enhanced')
        axes[1, i].axis('off')

    plt.tight_layout()
    plt.show()

# Compare some images
compare_images(x_train, x_train_enhanced)

# Plot training history
def plot_training_history(history):

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

    # Accuracy plot
    ax1.plot(history.history['accuracy'], label='Training Accuracy')
    ax1.plot(history.history['val_accuracy'], label='Validation Accuracy')
    ax1.set_title('Model Accuracy')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Accuracy')
    ax1.legend()
    ax1.grid(True)

    # Loss plot
    ax2.plot(history.history['loss'], label='Training Loss')
    ax2.plot(history.history['val_loss'], label='Validation Loss')
    ax2.set_title('Model Loss')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Loss')
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

plot_training_history(history)

# Test individual image enhancement
def test_single_image_enhancement():

    test_img = x_test[0]

    # Enhance single image
    enhanced_single = enhance_image_clarity(np.array([test_img]))[0]

    fig, axes = plt.subplots(1, 2, figsize=(10, 5))

    axes[0].imshow(test_img)
    axes[0].set_title('Original Image')
    axes[0].axis('off')

    axes[1].imshow(enhanced_single)
    axes[1].set_title('Enhanced Image')
    axes[1].axis('off')

    plt.tight_layout()
    plt.show()

    return enhanced_single

# Test single image enhancement
enhanced_test = test_single_image_enhancement()

# Save the model
model.save('cifar10_enhanced_cnn_model.h5')
print("Model saved as 'cifar10_enhanced_cnn_model.h5'")

# Function to predict on new images
def predict_image(model, image):

    if image.shape != (32, 32, 3):
        image = cv2.resize(image, (32, 32))

    # Enhance the image
    enhanced = enhance_image_clarity(np.array([image]))[0]

    # Make prediction
    prediction = model.predict(np.array([enhanced]))
    predicted_class = np.argmax(prediction)
    confidence = np.max(prediction)

    return class_names[predicted_class], confidence, enhanced

# Example usage of prediction function
sample_image = x_test[10]
pred_class, confidence, enhanced_img = predict_image(model, sample_image)
true_class = class_names[np.argmax(y_test[10])]

print(f"Prediction Example:")
print(f"True Class: {true_class}")
print(f"Predicted Class: {pred_class}")
print(f"Confidence: {confidence:.4f}")
```

## Analysis Summary


**Steps Undertaken and Rationale:**

1.  **Data Loading and Preparation:** The CIFAR-10 dataset was loaded, and the image data was normalized to a [0, 1] range. The labels were one-hot encoded.
2.  **Image Enhancement:** A custom function "enhance_image_clarity" was created and applied to both the training and test datasets. This function incorporates:
    *   **Sharpening:** To capture edges and fine details.
    *   **CLAHE (Contrast Limited Adaptive Histogram Equalization):** To improve local contrast, especially in images with varying lighting conditions.
    *   **Mild Gaussian Blur:** To reduce noise introduced by the enhancement process while preserving key features.
    The rationale was to investigate if providing the model with clearer, more detailed images would lead to better learning and classification accuracy.
3.  **Model Architecture:** A sequential CNN model was designed with multiple convolutional and pooling layers, batch normalization, and dropout for regularization. Global Average Pooling was used to reduce the number of parameters before the final dense layers.  
4.  **Model Compilation:** The model was compiled using the Adam optimizer with an exponential learning rate schedule. Categorical crossentropy was used as the loss function, appropriate for multi-class classification. An exponential decay schedule for the learning rate allows for larger steps in the beginning and smaller steps as training progresses.
5.  **Model Training:** The model was trained on the enhanced training data with early stopping based on validation loss. This prevents overfitting and selects the model with the best performance on unseen data.
6.  **Model Evaluation:** The trained model was evaluated on the enhanced test set to assess its generalization performance.
7.  **Performance Reporting:** A classification report was generated to provide detailed metrics (precision, recall, F1-score) for each class, offering a comprehensive view of the model's strengths and weaknesses across different categories.
8.  **Visualization:** Plots were generated to compare original and enhanced images visually and to track the training and validation accuracy and loss over epochs, providing insights into the training process.
10. **Prediction Example:** A function for predicting the class of a single new image was included.

**Model Performance:**

The model achieved a test accuracy of **0.8413** and a test loss of **0.4870**. The classification report indicates that the model performed well on many classes, but there were some challenges with classes like 'cat' and 'dog'. The image enhancement techniques appear to have contributed to a reasonably good performance on this dataset.
